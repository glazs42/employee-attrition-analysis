```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw_data.csv')
```

```python
# Drop columns that never change (zero analytical value)
df = df.drop(columns=['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'])
```

```python
# Confirming no duplicates
df = df.drop_duplicates()
```

```python
# Createing a numeric flag for easier aggregation later
df['AttritionFlag'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
# Saveing the cleaned version
df.to_csv('../data/cleaned_data.csv', index=False)
```

```python
attrition_rate = df['AttritionFlag'].mean() * 100
print(f"Overall attrition rate: {attrition_rate:.1f}%")
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Department', y='AttritionFlag', estimator=lambda x: sum(x)/len(x)*100)
plt.ylabel('Attrition Rate (%)')
plt.title('Attrition Rate by Department')
plt.savefig('../images/chart_attrition_by_department.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.barplot(data=df, x='OverTime', y='AttritionFlag', estimator=lambda x: sum(x)/len(x)*100)
plt.ylabel('Attrition Rate (%)')
plt.title('Attrition Rate: Overtime vs No Overtime')
plt.savefig('../images/chart_attrition_by_overtime.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income Distribution: Stayed vs Left')
plt.savefig('../images/chart_income_vs_attrition.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(12,10))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.savefig('../images/chart_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
sns.barplot(data=df, x='JobSatisfaction', y='AttritionFlag', estimator=lambda x: sum(x)/len(x)*100)
plt.ylabel('Attrition Rate (%)')
plt.title('Attrition Rate by Job Satisfaction Level (1=Low, 4=High)')
plt.savefig('../images/chart_satisfaction_vs_attrition.png', dpi=150, bbox_inches='tight')
plt.show()
```
